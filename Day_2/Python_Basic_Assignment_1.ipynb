## H1 — Subtraction, Multiplication and Division
Write a program to perform subtraction, multiplication and division.

```python
# H1: Perform subtraction, multiplication, and division

# Take two numbers from the user
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Perform the required arithmetic operations
subtraction = num1 - num2
multiplication = num1 * num2

# Division by zero is not allowed, so check before dividing
if num2 != 0:
    division = num1 / num2
    print("Subtraction:", subtraction)
    print("Multiplication:", multiplication)
    print("Division:", division)
else:
    print("Subtraction:", subtraction)
    print("Multiplication:", multiplication)
    print("Division is not possible because the second number is zero.")
```

## H5 — Average of Three Numbers
Calculate and print the average of three numbers, rounded to two decimal places.

```python
# H5: Calculate the average of three numbers

num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
num3 = float(input("Enter the third number: "))

# Calculate the average
average = (num1 + num2 + num3) / 3

# Round the result to two decimal places
print("Average:", round(average, 2))
```

## H6 — Simple Calculator
Create a calculator for addition, subtraction, multiplication, and division based on user input.

```python
# H6: Simple calculator

num1 = float(input("Enter the first number: "))
operator = input("Enter an operator (+, -, *, /): ")
num2 = float(input("Enter the second number: "))

# Select the operation based on the user's operator
if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 == 0:
        print("Error: Division by zero is not allowed.")
        result = None
    else:
        result = num1 / num2
else:
    print("Invalid operator.")
    result = None

if result is not None:
    print("Result:", result)
```

## H7 — Vowel or Consonant
Ask the user for a single character and determine whether it is a vowel or consonant.

```python
# H7: Check whether a character is a vowel or consonant

character = input("Enter a single character: ").lower()

# First make sure the user entered exactly one alphabetic character
if len(character) != 1 or not character.isalpha():
    print("Please enter exactly one alphabetic character.")
elif character in "aeiou":
    print("The character is a vowel.")
else:
    print("The character is a consonant.")
```

## H8 — Fahrenheit and Celsius Conversion
Convert temperatures between Fahrenheit and Celsius based on user input.

```python
# H8: Convert temperature between Celsius and Fahrenheit

choice = input("Enter C to convert Celsius to Fahrenheit, or F to convert Fahrenheit to Celsius: ").upper()
temperature = float(input("Enter the temperature: "))

if choice == "C":
    # Celsius to Fahrenheit: (C × 9/5) + 32
    fahrenheit = (temperature * 9 / 5) + 32
    print("Temperature in Fahrenheit:", round(fahrenheit, 2))
elif choice == "F":
    # Fahrenheit to Celsius: (F - 32) × 5/9
    celsius = (temperature - 32) * 5 / 9
    print("Temperature in Celsius:", round(celsius, 2))
else:
    print("Invalid choice. Please enter C or F.")
```

## H9 — Find the Largest of Three Numbers
Ask the user for three numbers and determine which one is the largest.

```python
# H9: Find the largest among three numbers

num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
num3 = float(input("Enter the third number: "))

# Compare the three numbers
if num1 >= num2 and num1 >= num3:
    largest = num1
elif num2 >= num1 and num2 >= num3:
    largest = num2
else:
    largest = num3

print("The largest number is:", largest)
```

## H10 — Print Even Numbers from 1 to N
Print all even numbers between 1 and N using a for loop.

```python
# H10: Print all even numbers from 1 to N

n = int(input("Enter the value of N: "))

print("Even numbers between 1 and", n, ":")

# range(2, n + 1, 2) generates even numbers starting from 2
for number in range(2, n + 1, 2):
    print(number, end=" ")
```

## H11 — Sum of Numbers Divisible by 3 or 5
Calculate the sum of all numbers from 1 to N that are divisible by 3 or 5.

```python
# H11: Sum numbers divisible by 3 or 5

n = int(input("Enter the value of N: "))
total = 0

# Check every number from 1 to N
for number in range(1, n + 1):
    if number % 3 == 0 or number % 5 == 0:
        total += number

print("Sum of numbers divisible by 3 or 5:", total)
```

## H12 — Multiplication Table
Print the multiplication table of n from 1 to 10 using a for loop.

```python
# H12: Print the multiplication table of n

n = int(input("Enter a number: "))

print(f"Multiplication table of {n}:")

# Generate the table from 1 to 10
for i in range(1, 11):
    print(f"{n} x {i} = {n * i}")
```
